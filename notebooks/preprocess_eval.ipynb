```python
import os
from pathlib import Path
import pandas as pd


def save_images_by_label(df, output_dir="output_images"):
    output_base = Path(output_dir)

    for idx, row in df.iterrows():
        img_dict = row["image"]

        # Ensure we have valid data
        if not isinstance(img_dict, dict) or "bytes" not in img_dict:
            continue

        byte_data = img_dict["bytes"]
        path_str = img_dict.get("path", f"image_{idx}.jpg")

        # Get filename (e.g., "Mountain_33.jpg")
        filename = Path(path_str).name

        # Extract label before the last underscore (e.g., "Mountain_33.jpg" -> "Mountain")
        # If no underscore exists, default to the filename stem
        if "_" in filename:
            label = filename.rsplit("_", -1)[0]
        elif "-" in filename:
            label = filename.rsplit("-", -1)[0]
        else:
            label = Path(filename).stem

        # Create target subfolder directory (e.g., "output_images/Mountain")
        label_folder = output_base / label
        label_folder.mkdir(parents=True, exist_ok=True)

        # Write binary bytes to disk
        file_path = label_folder / filename
        with open(file_path, "wb") as f:
            f.write(byte_data)


df = pd.read_parquet(rf"C:\Users\sus14836\Downloads\train-00005-of-00006-1e9efe76b85e7e61.parquet")
save_images_by_label(df, output_dir=rf"E:\Data\vlm\eval\CLRS")
```
