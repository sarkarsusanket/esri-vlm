```python
from ultralytics import YOLO

# Initialize a YOLO-World model
model = YOLO("yolov8m-world.pt")  # or choose yolov8m/l-world.pt

# Define custom classes
model.set_classes(["pools"])

# Execute prediction for specified categories on an image
results = model.predict("D:\Code\esri-vlm\data\image.png")

# Show results
results[0].show()
```

```python
from ultralytics import YOLO

# Initialize a YOLO-World model
model = YOLO("yolov8s-world.pt")  # or choose yolov8m/l-world.pt

# Define custom classes
model.set_classes(["cars"])

# Execute prediction for specified categories on an image
results = model.predict("D:\Code\esri-vlm\data\image2.png")

# Show results
results[0].show()
```

```python
from ultralytics import YOLOWorld

# Initialize a YOLO-World model
model = YOLOWorld("yolov8s-world.pt")  # or select yolov8m/l-world.pt for different sizes

# Execute inference with the YOLOv8s-world model on the specified image
results = model.predict("D:\Code\esri-vlm\data\image2.png")

# Show results
results[0].show()
```
